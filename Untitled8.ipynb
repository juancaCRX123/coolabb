Las arquitecturas de redes neuronales más avanzadas utilizan funciones de activación como GELU, Swish y Mish. Para mantener el flujo de datos estable durante el entrenamiento se recurre a la normalización, un proceso donde primero se calcula la media y la desviación estándar de cada mini-lote para centrar las entradas en cero, y después la señal se escala y se desplaza aplicando los parámetros aprendidos gamma y beta. Por otro lado, el aprendizaje transferido permite aprovechar redes profundas ya entrenadas para resolver nuevas tareas con un menor consumo de recursos, congelando las capas base del modelo y ajustando únicamente las capas superiores. Esta metodología basada en el preentrenamiento y el ajuste fino terminó siendo un factor clave que impulsó el renacimiento de las redes neuronales.
