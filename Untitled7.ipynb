El material explica la lógica detrás de los métodos de ensamblaje en aprendizaje automático para mejorar el rendimiento de las predicciones. Por un lado, presenta los clasificadores por votación, donde modelos de distinta naturaleza —como regresión logística, SVM o Random Forest— analizan el mismo conjunto de datos para llegar a un consenso. Esta estrategia es tan efectiva que permite transformar predictores individuales con un acierto modesto, de apenas un 51%, en un sistema conjunto mucho más preciso; para ello, se apoya en votaciones directas donde cada modelo vale un voto (hard voting) o en el promedio de sus probabilidades asignadas (soft voting). Por otro lado, aborda el método de bagging, el cual divide los datos iniciales mediante un muestreo aleatorio con reemplazo para entrenar múltiples clasificadores en paralelo, logrando así reducir la varianza general y aportar mayor solidez al modelo final.

Este bloque aborda técnicas avanzadas de ensamblaje comenzando con los bosques aleatorios, los cuales introducen aleatoriedad en las variables mediante estrategias de muestreo como los parches aleatorios (que seleccionan tanto instancias como características) y los subespacios aleatorios (que seleccionan únicamente características). Posteriormente, presenta los algoritmos de boosting, centrados en la corrección secuencial del error; por ejemplo, en AdaBoost se entrena un modelo inicial, se le otorga mayor peso a los fallos para enfocar el siguiente clasificador en ellos y se repite este ciclo para ajustar los errores de forma progresiva. Finalmente, introduce el concepto de mezclador (blender), una técnica donde un modelo final se entrena específicamente para tomar las predicciones generadas por los clasificadores previos y combinarlas en un resultado único.
