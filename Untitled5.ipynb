Las Máquinas de Vectores de Soporte, o SVM, son modelos bastante versátiles que sirven tanto para clasificación lineal o no lineal como para regresión y detección. En los problemas sencillos, el modelo busca separar las clases mediante un margen lo más amplio posible, apoyándose en los puntos ubicados justo en el límite, conocidos como vectores de soporte. Sin embargo, cuando los datos no son separables de forma directa, se puede recurrir al mapeo polinomial, un proceso donde se identifican los datos no lineales, se transforman agregando potencias a sus características para llevarlos a dimensiones superiores y finalmente se entrena un modelo lineal en ese nuevo espacio.

Para evitar el costo computacional de crear todas esas variables manualmente, entra en juego el truco del kernel, una técnica matemática que logra el mismo resultado sin tener que añadirlas de forma explícita. Mediante un kernel polinomial se pueden definir fronteras de decisión complejas según el grado que se utilice. Asimismo, existen alternativas como la función RBF Gaussiana, la cual crea características basadas en la similitud entre puntos. En este tipo de modelos, ajustar los parámetros Gamma y C es fundamental para definir el alcance de influencia de los datos y regular qué tan flexible o ajustada será la frontera final.

En el caso de la regresión con SVM, el enfoque cambia por completo: en lugar de dejar los datos fuera del margen para separarlos, el objetivo pasa a ser meter la mayor cantidad posible de puntos dentro de una "calle" delimitada por el parámetro épsilon. Cuando la relación entre las variables es más compleja, esta misma idea se adapta aplicando regresión no lineal mediante un kernel polinomial y ajustando la regularización.

A nivel matemático interno, la clave para lograr que esa frontera tenga un margen mucho más amplio consiste en minimizar los pesos del modelo. Además, el costo computacional varía según la clase que se utilice en la práctica: herramientas como LinearSVC o SGDClassifier son eficientes y escalan de forma lineal sin usar kernels, mientras que SVC sí permite usarlos pero requiere más tiempo de cómputo a medida que crece el volumen de datos.
