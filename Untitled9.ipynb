Las redes neuronales recurrentes están diseñadas para procesar secuencias de cualquier longitud preservando un estado oculto a través de conexiones internas. Al desplegar estas neuronas a lo largo del tiempo, la red conecta cada entrada con las salidas calculadas y la información acumulada en los pasos previos. Su entrenamiento se basa en la retropropagación a través del tiempo (BPTT), proceso compuesto por un pase adelante para procesar la secuencia, la evaluación de la salida mediante una función de pérdida, la propagación de gradientes de error hacia atrás mediante la regla de la cadena y el ajuste de parámetros por descenso de gradiente. Esta estructura resulta fundamental para el pronóstico de series temporales, permitiendo trabajar con modelos univariantes que emplean una sola variable (como autobuses) o multivariantes que integran múltiples factores (como autobuses, trenes y el día de la semana), al tiempo que abre paso a resolver la memoria a corto plazo usando celdas LSTM o GRU y enfoques alternativos como WaveNet o CNNs 1D.

Las redes neuronales recurrentes procesan secuencias de cualquier longitud mediante conexiones internas que mantienen un estado oculto a lo largo del tiempo. Al desplegar la red temporalmente, el entrenamiento se realiza mediante retropropagación a través del tiempo (BPTT), procedimiento que ejecuta un pase hacia adelante, evalúa la función de pérdida, propaga los gradientes de error hacia atrás mediante la regla de la cadena y ajusta los parámetros con descenso de gradiente. Aunque este esquema resulta útil para el pronóstico de series temporales tanto univariantes como multivariantes, enfrenta el obstáculo crítico de la memoria a corto plazo, donde la red tiende a olvidar la información inicial. Para solucionar esta limitación se emplean celdas de memoria como las LSTM, que gestionan la información mediante compuertas de olvido, entrada y salida, o las GRU, que simplifican este mecanismo mediante un diseño de compuertas más compacto. Finalmente, para trabajar con secuencias especialmente largas de forma eficiente, se recurre a redes convolucionales 1D y a la arquitectura apilada de WaveNet, la cual utiliza capas con convoluciones dilatadas.
