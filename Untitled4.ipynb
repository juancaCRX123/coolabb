Esta clase explica cómo enseñar a una computadora a clasificar información tomando como ejemplo el dataset MNIST, donde el modelo aprende a identificar números escritos a mano. Para entender la lógica básica, empieza con un clasificador binario que únicamente responde si una imagen es o no es un número cinco. Esto ayuda a demostrar por qué medir solo la exactitud general es engañoso: un programa podría responder siempre que «no» y aparentar un porcentaje alto de éxito simplemente porque la mayoría de los números en el grupo no son un cinco.

Para medir los errores reales, la presentación introduce la matriz de confusión, que muestra en qué dígitos específicos se equivoca el sistema, y la curva ROC, que ayuda a encontrar el equilibrio entre detectar positivos y evitar falsos errores. Finalmente, el tema escala a la clasificación multiclase para reconocer todos los dígitos al mismo tiempo, concluyendo con la decisión técnica más importante de cualquier proyecto: definir si conviene priorizar la precisión (asegurarte de no dar falsos positivos) o la exhaustividad (evitar que se te escape cualquier caso real).
